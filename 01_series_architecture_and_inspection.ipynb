# Pandas Series: Architecture & Attribute Inspection

```python
import pandas as pd
```

## 1. Creating a Pandas Series

```python
# Initialize a 1D (One-Dimensional) Pandas Series with custom index labels
s = pd.Series(
    data=[10, 20, 20, 30, 40],       # Values (contains duplicate 20)
    index=['a', 'b', 'c', 'd', 'e']  # Custom string labels instead of default 0-4
)

# Display the Series and its inferred data type (int64)
print(s)
```

## 2. Inspecting Series Attributes & Methods


```python
# Inspect core attributes of the Series (values, index labels, and data type)
print("Values:", s.values)  # Returns underlying data as a NumPy array
print("Index: ", s.index)   # Returns the Index object holding row labels
print("Dtype: ", s.dtype)   # Returns the data type of the stored elements (int64)

# Preview top and bottom elements
print(s.head())  # Default: returns first 5 rows (all rows here)
print(s.tail())  # Default: returns last 5 rows (all rows here)

# Count how many times each unique value appears
print(s.value_counts())  # Shows 20 appears twice; 10, 30, 40 appear once

```
